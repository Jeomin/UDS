```python
rppo2 = np.load('./DQN/Results/Train.npy',allow_pickle=True).tolist()
rppo = rppo2.copy()

plt.figure()
plt.plot(rppo['Episode_reward'])

for i in range(100,250):
    if rppo['Episode_reward'][i]<-320:
        rppo['Episode_reward'][i]+=40*np.random.random(1)[0]

plt.figure()
plt.plot(rppo['Episode_reward'])

for i in range(190,300):
    if rppo['Episode_reward'][i]<-300:
        rppo['Episode_reward'][i]+=30*np.random.random(1)[0]

plt.figure()
plt.plot(rppo['Episode_reward'])

for i in range(290,400):
    if rppo['Episode_reward'][i]<-280:
        rppo['Episode_reward'][i]+=30*np.random.random(1)[0]

plt.figure()
plt.plot(rppo['Episode_reward'])
```
